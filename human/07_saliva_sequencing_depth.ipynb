# Saliva sequencing depth

Sequencing methods. The stored result matched the saliva mean and sample SD. This notebook expects the original FASTQ file set and retains its OM-501 filename filter.

```python
from pathlib import Path

# Jupyter normally starts the kernel in the notebook directory.
here = Path.cwd()
repo_root = next(parent for parent in (here, *here.parents)
                 if (parent / "requirements.txt").is_file()
                 and (parent / "mouse").is_dir() and (parent / "human").is_dir())
data_dir = repo_root / "data"
output_dir = repo_root / "outputs" / 'human/07_saliva_sequencing_depth'
output_dir.mkdir(parents=True, exist_ok=True)
notebook_output = output_dir
```

## Select the original saliva FASTQ files

```python
import glob

folder = data_dir / "sequencing" / "LEVM009"

# Get all fastq.gz files
fastq_files = glob.glob(str(folder / "*.fastq.gz"))

# Filter to include only files containing "OM-501"
fastq_files = [f for f in fastq_files if "OM-501" in f]

print("Files found:", len(fastq_files))
fastq_files[:5]
```

## Count records and calculate mean and sample SD

```python
from Bio import SeqIO
import gzip

read_counts = []

for f in fastq_files:
    with gzip.open(f, "rt") as handle:  # text mode
        count = sum(1 for _ in SeqIO.parse(handle, "fastq"))
    
    read_counts.append({
        "file": Path(f).name,
        "reads": count
    })
```

```python
import pandas as pd
import numpy as np

df = pd.DataFrame(read_counts)

mean_reads = df["reads"].mean()
std_reads = df["reads"].std()

print("Mean reads:", int(mean_reads))
print("Std dev:", int(std_reads))
```
